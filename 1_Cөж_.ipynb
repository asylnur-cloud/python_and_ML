### Тапсырма 19. Баллды бағаға айналдыру

`score` айнымалысындағы балл бойынша бағаны анықтаңыз: 90 және одан жоғары — "A", 75–89 — "B", 50–74 — "C", 50-ден төмен — "F". Балл 0-ден кіші немесе 100-ден үлкен болса, "Қате балл" деп шығарыңыз.

````python
# Шешіміңізді осында жазыңыз
score=90

if score < 0 or score > 100:
   grade = "Қате балл"
elif score>=90:
   grade="A"
elif score>=75:
  grade="B"
elif score>=50:
  grade="C"
else:
  grade="F"

print(f"Балл: {score}, баға: {grade}")
````

### Тапсырма 20. Кібісе жыл

`year` айнымалысындағы жылдың кібісе жыл екенін анықтаңыз. Жыл 4-ке бөлінсе және 100-ге бөлінбесе, немесе 400-ге бөлінсе, ол кібісе жыл болып саналады. `2024`, `1900` және `2000` жылдарымен тексеріңіз.

**Күтілетін нәтиже:**

```
2024 - кібісе жыл
1900 - кібісе жыл емес
2000 - кібісе жыл
```

````python
# Шешіміңізді осында жазыңыз
years=[2024,1900,2000]

for year in years:

  if (year %4==0 and year %100!= 0) or year %400 == 0:
   print(f"{year}- кібісе жыл")
  else:
   print(f"{year} - кібісе жыл емес")
````

### Тапсырма 22. Үшбұрыштың түрі

Үшбұрыштың үш қабырғасының ұзындығы `a`, `b`, `c` берілген. Алдымен мұндай үшбұрыштың болуы мүмкін екенін тексеріңіз: кез келген екі қабырғаның қосындысы үшінші қабырғадан үлкен болуы керек. Үшбұрыш бар болса, оның тең қабырғалы, теңбүйірлі немесе әртүрлі қабырғалы екенін анықтаңыз.

````python
# Шешіміңізді осында жазыңыз
a=3
b=4
c=6

if a+b>c or b+c>a and a+c>b:
  if a==b==c:
    print("тең қабырғалы үшбұрыш")
  elif a==b and b==c and c==b:
    print("теңбүйірлі үшбұрыш")
  else:
    print("әртүрлі қабырғалы үшбұрыш")
````

### Тапсырма 23. Қарапайым калькулятор

````python
# Шешіміңізді осында жазыңыз
a=28
b=12
operation=("+")

if operation=="+":
  print(a+b)
elif operation=="-":
  print(a-b)
elif operation=="*":
  print(a*b)
elif operation=="/":
  if b!=0:
   print(a/b)
  else:
   print("нөлге бөлуге болмайды")
else:
  print("белгісіз амал")
````

```
is_palindrome("Python") -> False
```

````python
# Шешіміңізді осында жазыңыз
def is_palindrome(text):
  cleaned=text.lower() .replace(" ","")
  return cleaned ==cleaned[::-1]

print(is_palindrome("Level"))                         # True
print(is_palindrome("Was it a car or a cat I saw"))   # True
print(is_palindrome("Python"))                        # False
````
